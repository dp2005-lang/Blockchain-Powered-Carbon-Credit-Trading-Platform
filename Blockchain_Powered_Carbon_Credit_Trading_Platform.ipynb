```python
# ================================================================
# BLOCKCHAIN-POWERED CARBON CREDIT TRADING PLATFORM
# Google Colab - SINGLE CELL COMPLETE SIMULATION
# ================================================================

# -----------------------------
# 1. INSTALL / IMPORT LIBRARIES
# -----------------------------
import hashlib
import time
import json
import random
from datetime import datetime
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Image

print("=" * 75)
print("BLOCKCHAIN-POWERED CARBON CREDIT TRADING PLATFORM")
print("Google Colab Blockchain Simulation")
print("=" * 75)


# ================================================================
# 2. SIMULATED BLOCKCHAIN
# ================================================================

class CarbonCreditBlockchain:

    def __init__(self):

        # Simulated wallets
        self.wallets = {
            "ADMIN": "0xADMIN001",
            "ISSUER": "0xISSUER002",
            "SELLER": "0xSELLER003",
            "BUYER": "0xBUYER004"
        }

        # Authorized issuers
        self.authorized_issuers = set()

        # Carbon credits
        self.credits = {}

        # Marketplace listings
        self.listings = {}

        # Blockchain event log
        self.events = []

        # Transaction list
        self.transactions = []

        # Counters
        self.tx_counter = 0

        # Supply statistics
        self.total_issued = 0
        self.total_retired = 0
        self.total_traded = 0

        # Status values
        self.STATUS_ACTIVE = "ACTIVE"
        self.STATUS_LISTED = "LISTED"
        self.STATUS_TRANSFERRED = "TRANSFERRED"
        self.STATUS_RETIRED = "RETIRED"


    # ============================================================
    # BLOCKCHAIN TRANSACTION HASH
    # ============================================================

    def create_transaction(self, action, data):

        self.tx_counter += 1

        raw = (
            str(self.tx_counter)
            + action
            + json.dumps(data, sort_keys=True)
            + str(time.time())
        )

        tx_hash = "0x" + hashlib.sha256(
            raw.encode()
        ).hexdigest()[:24]

        tx = {
            "tx_hash": tx_hash,
            "action": action,
            "timestamp": datetime.now().strftime(
                "%Y-%m-%d %H:%M:%S"
            )
        }

        self.transactions.append(tx)

        return tx_hash


    # ============================================================
    # EVENT LOGGER
    # ============================================================

    def emit_event(self, event_name, data):

        event = {
            "event": event_name,
            "data": data,
            "timestamp": datetime.now().strftime(
                "%Y-%m-%d %H:%M:%S"
            )
        }

        self.events.append(event)


    # ============================================================
    # REGISTER ISSUER
    # ============================================================

    def register_issuer(self, issuer):

        if issuer not in self.wallets.values():
            raise Exception("Invalid issuer address")

        if issuer in self.authorized_issuers:
            raise Exception("Issuer already registered")

        self.authorized_issuers.add(issuer)

        tx_hash = self.create_transaction(
            "REGISTER_ISSUER",
            {"issuer": issuer}
        )

        self.emit_event(
            "IssuerRegistered",
            {"issuer": issuer}
        )

        print("\n[ISSUER REGISTRATION]")
        print("Issuer:", issuer)
        print("Status: AUTHORIZED")
        print("Transaction:", tx_hash)

        return tx_hash


    # ============================================================
    # ISSUE CARBON CREDIT
    # ============================================================

    def issue_carbon_credit(
        self,
        credit_id,
        project_name,
        project_type,
        country,
        vintage_year,
        tonnes,
        issuer,
        owner,
        metadata_hash
    ):

        # Security checks
        if issuer not in self.authorized_issuers:
            raise Exception(
                "Only authorized issuer can issue credits"
            )

        if credit_id in self.credits:
            raise Exception(
                "Credit ID already exists"
            )

        if tonnes <= 0:
            raise Exception(
                "Tonnes CO2e must be greater than zero"
            )

        if owner not in self.wallets.values():
            raise Exception(
                "Invalid owner address"
            )

        # Carbon credit structure
        credit = {

            "credit_id": credit_id,

            "project_name": project_name,

            "project_type": project_type,

            "country": country,

            "vintage_year": vintage_year,

            "tonnes_co2e": tonnes,

            "issuer": issuer,

            "owner": owner,

            "metadata_hash": metadata_hash,

            "status": self.STATUS_ACTIVE,

            "created_at":
                datetime.now().strftime(
                    "%Y-%m-%d %H:%M:%S"
                ),

            "retired_at": None,

            "retirement_reason": None
        }

        self.credits[credit_id] = credit

        self.total_issued += tonnes

        tx_hash = self.create_transaction(
            "ISSUE_CREDIT",
            credit
        )

        self.emit_event(
            "CreditIssued",
            {
                "credit_id": credit_id,
                "owner": owner,
                "tonnes": tonnes
            }
        )

        print("\n[CREDIT ISSUED]")
        print("Credit ID:", credit_id)
        print("Project:", project_name)
        print("CO2e:", tonnes, "tonnes")
        print("Owner:", owner)
        print("Status: ACTIVE")
        print("Transaction:", tx_hash)

        return tx_hash


    # ============================================================
    # GET CREDIT DETAILS
    # ============================================================

    def get_credit_details(self, credit_id):

        if credit_id not in self.credits:
            raise Exception("Credit does not exist")

        return self.credits[credit_id]


    # ============================================================
    # LIST CREDIT FOR SALE
    # ============================================================

    def list_credit_for_sale(
        self,
        credit_id,
        seller,
        price
    ):

        if credit_id not in self.credits:
            raise Exception("Credit does not exist")

        credit = self.credits[credit_id]

        if credit["owner"] != seller:
            raise Exception(
                "Only current owner can list credit"
            )

        if credit["status"] == self.STATUS_RETIRED:
            raise Exception(
                "Retired credit cannot be listed"
            )

        if price <= 0:
            raise Exception(
                "Price must be greater than zero"
            )

        if credit_id in self.listings:
            raise Exception(
                "Credit is already listed"
            )

        listing = {

            "credit_id": credit_id,

            "seller": seller,

            "price": price,

            "active": True,

            "listed_at":
                datetime.now().strftime(
                    "%Y-%m-%d %H:%M:%S"
                )
        }

        self.listings[credit_id] = listing

        credit["status"] = self.STATUS_LISTED

        tx_hash = self.create_transaction(
            "LIST_CREDIT",
            listing
        )

        self.emit_event(
            "CreditListed",
            listing
        )

        print("\n[MARKETPLACE LISTING]")
        print("Credit:", credit_id)
        print("Seller:", seller)
        print("Price:", price, "TEST ETH")
        print("Status: LISTED")
        print("Transaction:", tx_hash)

        return tx_hash


    # ============================================================
    # BUY CREDIT
    # ============================================================

    def buy_credit(
        self,
        credit_id,
        buyer
    ):

        if credit_id not in self.listings:
            raise Exception(
                "Credit is not listed"
            )

        listing = self.listings[credit_id]

        if not listing["active"]:
            raise Exception(
                "Listing is closed"
            )

        credit = self.credits[credit_id]

        if credit["status"] == self.STATUS_RETIRED:
            raise Exception(
                "Retired credit cannot be purchased"
            )

        seller = listing["seller"]

        # Ownership transfer
        credit["owner"] = buyer

        credit["status"] = self.STATUS_TRANSFERRED

        # Close listing
        listing["active"] = False

        self.total_traded += credit["tonnes_co2e"]

        tx_hash = self.create_transaction(
            "BUY_CREDIT",
            {
                "credit_id": credit_id,
                "seller": seller,
                "buyer": buyer,
                "price": listing["price"]
            }
        )

        self.emit_event(
            "CreditPurchased",
            {
                "credit_id": credit_id,
                "seller": seller,
                "buyer": buyer
            }
        )

        self.emit_event(
            "CreditTransferred",
            {
                "credit_id": credit_id,
                "from": seller,
                "to": buyer
            }
        )

        print("\n[CREDIT PURCHASED]")
        print("Credit:", credit_id)
        print("Seller:", seller)
        print("Buyer:", buyer)
        print("Price:", listing["price"], "TEST ETH")
        print("New Owner:", buyer)
        print("Transaction:", tx_hash)

        return tx_hash


    # ============================================================
    # DIRECT TRANSFER
    # ============================================================

    def transfer_credit(
        self,
        credit_id,
        sender,
        receiver
    ):

        if credit_id not in self.credits:
            raise Exception("Credit does not exist")

        credit = self.credits[credit_id]

        if credit["owner"] != sender:
            raise Exception(
                "Only current owner can transfer"
            )

        if credit["status"] == self.STATUS_RETIRED:
            raise Exception(
                "Retired credit cannot be transferred"
            )

        if receiver not in self.wallets.values():
            raise Exception(
                "Invalid receiver"
            )

        credit["owner"] = receiver

        credit["status"] = self.STATUS_TRANSFERRED

        tx_hash = self.create_transaction(
            "TRANSFER_CREDIT",
            {
                "credit_id": credit_id,
                "from": sender,
                "to": receiver
            }
        )

        self.emit_event(
            "CreditTransferred",
            {
                "credit_id": credit_id,
                "from": sender,
                "to": receiver
            }
        )

        return tx_hash


    # ============================================================
    # RETIRE CREDIT
    # ============================================================

    def retire_credit(
        self,
        credit_id,
        owner,
        reason
    ):

        if credit_id not in self.credits:
            raise Exception(
                "Credit does not exist"
            )

        credit = self.credits[credit_id]

        if credit["owner"] != owner:
            raise Exception(
                "Only current owner can retire"
            )

        if credit["status"] == self.STATUS_RETIRED:
            raise Exception(
                "Credit is already retired"
            )

        # Retirement is irreversible
        credit["status"] = self.STATUS_RETIRED

        credit["retired_at"] = (
            datetime.now().strftime(
                "%Y-%m-%d %H:%M:%S"
            )
        )

        credit["retirement_reason"] = reason

        self.total_retired += credit["tonnes_co2e"]

        tx_hash = self.create_transaction(
            "RETIRE_CREDIT",
            {
                "credit_id": credit_id,
                "owner": owner,
                "reason": reason
            }
        )

        self.emit_event(
            "CreditRetired",
            {
                "credit_id": credit_id,
                "owner": owner,
                "tonnes": credit["tonnes_co2e"],
                "reason": reason
            }
        )

        print("\n[CREDIT RETIRED]")
        print("Credit:", credit_id)
        print("Owner:", owner)
        print("Reason:", reason)
        print("Status: RETIRED")
        print("Transaction:", tx_hash)

        return tx_hash


    # ============================================================
    # SUPPLY METRICS
    # ============================================================

    def supply_metrics(self):

        active = 0
        listed = 0
        retired = 0

        for credit in self.credits.values():

            if credit["status"] == "ACTIVE":
                active += credit["tonnes_co2e"]

            elif credit["status"] == "LISTED":
                listed += credit["tonnes_co2e"]

            elif credit["status"] == "RETIRED":
                retired += credit["tonnes_co2e"]

            elif credit["status"] == "TRANSFERRED":
                active += credit["tonnes_co2e"]

        return {
            "Total Issued": self.total_issued,
            "Active": active,
            "Listed": listed,
            "Retired": retired,
            "Traded": self.total_traded
        }


# ================================================================
# 3. CREATE BLOCKCHAIN
# ================================================================

blockchain = CarbonCreditBlockchain()

ADMIN = blockchain.wallets["ADMIN"]
ISSUER = blockchain.wallets["ISSUER"]
SELLER = blockchain.wallets["SELLER"]
BUYER = blockchain.wallets["BUYER"]


# ================================================================
# 4. REGISTER AUTHORIZED ISSUER
# ================================================================

blockchain.register_issuer(ISSUER)


# ================================================================
# 5. CREATE SIMULATED CARBON CREDIT
# ================================================================

metadata = {
    "project_name": "Solar Energy Project",
    "project_type": "Renewable Energy",
    "country": "India",
    "vintage_year": 2026,
    "verification_status":
        "SIMULATED - NOT OFFICIALLY VERIFIED",
    "co2e_quantity": 10,
    "document_hash":
        "SIMULATED_DOCUMENT_HASH_001"
}

metadata_string = json.dumps(
    metadata,
    sort_keys=True
)

metadata_hash = hashlib.sha256(
    metadata_string.encode()
).hexdigest()


blockchain.issue_carbon_credit(

    credit_id="CC-2026-001",

    project_name="Solar Energy Project",

    project_type="Renewable Energy",

    country="India",

    vintage_year=2026,

    tonnes=10,

    issuer=ISSUER,

    owner=SELLER,

    metadata_hash=metadata_hash
)


# ================================================================
# 6. VERIFY CREDIT
# ================================================================

print("\n" + "=" * 75)
print("CREDIT DETAILS")
print("=" * 75)

credit = blockchain.get_credit_details(
    "CC-2026-001"
)

for key, value in credit.items():
    print(f"{key:25} : {value}")


# ================================================================
# 7. LIST CREDIT
# ================================================================

blockchain.list_credit_for_sale(
    credit_id="CC-2026-001",
    seller=SELLER,
    price=0.05
)


# ================================================================
# 8. BUY CREDIT
# ================================================================

blockchain.buy_credit(
    credit_id="CC-2026-001",
    buyer=BUYER
)


# ================================================================
# 9. VERIFY OWNERSHIP
# ================================================================

print("\n" + "=" * 75)
print("OWNERSHIP AFTER PURCHASE")
print("=" * 75)

print(
    "Current Owner:",
    blockchain.credits["CC-2026-001"]["owner"]
)

print(
    "Status:",
    blockchain.credits["CC-2026-001"]["status"]
)


# ================================================================
# 10. RETIRE CREDIT
# ================================================================

blockchain.retire_credit(

    credit_id="CC-2026-001",

    owner=BUYER,

    reason="Corporate Net-Zero Demonstration"
)


# ================================================================
# 11. SECURITY TESTS
# ================================================================

print("\n" + "=" * 75)
print("SECURITY VALIDATION")
print("=" * 75)

security_results = []


# Test 1: Retired transfer
try:

    blockchain.transfer_credit(
        "CC-2026-001",
        BUYER,
        SELLER
    )

    security_results.append(
        ["Retired Credit Transfer",
         "FAILED",
         "Transfer unexpectedly allowed"]
    )

except Exception as e:

    security_results.append(
        ["Retired Credit Transfer",
         "PASSED",
         str(e)]
    )


# Test 2: Retired listing
try:

    blockchain.list_credit_for_sale(
        "CC-2026-001",
        BUYER,
        0.1
    )

    security_results.append(
        ["Retired Credit Listing",
         "FAILED",
         "Listing unexpectedly allowed"]
    )

except Exception as e:

    security_results.append(
        ["Retired Credit Listing",
         "PASSED",
         str(e)]
    )


# Test 3: Unauthorized issuance
try:

    blockchain.issue_carbon_credit(
        credit_id="CC-HACK-001",
        project_name="Fake Project",
        project_type="Unknown",
        country="India",
        vintage_year=2026,
        tonnes=100,
        issuer=BUYER,
        owner=BUYER,
        metadata_hash="fake"
    )

    security_results.append(
        ["Unauthorized Issuance",
         "FAILED",
         "Unauthorized issuer accepted"]
    )

except Exception as e:

    security_results.append(
        ["Unauthorized Issuance",
         "PASSED",
         str(e)]
    )


# Test 4: Zero tonnes
try:

    blockchain.issue_carbon_credit(
        credit_id="CC-ZERO",
        project_name="Zero Project",
        project_type="Test",
        country="India",
        vintage_year=2026,
        tonnes=0,
        issuer=ISSUER,
        owner=SELLER,
        metadata_hash="zero"
    )

    security_results.append(
        ["Zero Tonnes Validation",
         "FAILED",
         "Zero credit accepted"]
    )

except Exception as e:

    security_results.append(
        ["Zero Tonnes Validation",
         "PASSED",
         str(e)]
    )


security_df = pd.DataFrame(
    security_results,
    columns=[
        "Security Test",
        "Result",
        "Validation"
    ]
)

display(security_df)


# ================================================================
# 12. SUPPLY METRICS
# ================================================================

metrics = blockchain.supply_metrics()

print("\n" + "=" * 75)
print("LIVE BLOCKCHAIN SUPPLY METRICS")
print("=" * 75)

for key, value in metrics.items():
    print(f"{key:20} : {value} tCO2e")


# ================================================================
# 13. EVENT LOG
# ================================================================

print("\n" + "=" * 75)
print("BLOCKCHAIN AUDIT EVENTS")
print("=" * 75)

for i, event in enumerate(
    blockchain.events,
    start=1
):

    print(
        f"{i}. {event['event']} "
        f"| {event['timestamp']}"
    )


# ================================================================
# 14. TRANSACTION HASHES
# ================================================================

print("\n" + "=" * 75)
print("TRANSACTION HISTORY")
print("=" * 75)

for tx in blockchain.transactions:

    print(
        tx["action"],
        "->",
        tx["tx_hash"]
    )


# ================================================================
# 15. CREATE RESULT DATAFRAME
# ================================================================

result_data = {

    "Project":
        ["Solar Energy Project"],

    "Credit ID":
        ["CC-2026-001"],

    "Vintage":
        [2026],

    "CO2e Issued":
        [10],

    "Seller":
        [SELLER],

    "Buyer":
        [BUYER],

    "Final Owner":
        [
            blockchain.credits[
                "CC-2026-001"
            ]["owner"]
        ],

    "Final Status":
        [
            blockchain.credits[
                "CC-2026-001"
            ]["status"]
        ],

    "Retirement":
        ["YES"],

    "Verification":
        ["SIMULATED"]
}

result_df = pd.DataFrame(result_data)

display(result_df)


# ================================================================
# 16. CREATE VISUAL DASHBOARD
# ================================================================

plt.figure(figsize=(16, 10))

plt.suptitle(
    "BLOCKCHAIN-POWERED CARBON CREDIT TRADING PLATFORM",
    fontsize=20,
    fontweight="bold"
)


# ------------------------------------------------
# Supply Chart
# ------------------------------------------------

plt.subplot(2, 2, 1)

labels = [
    "Total Issued",
    "Active",
    "Traded",
    "Retired"
]

values = [
    metrics["Total Issued"],
    metrics["Active"],
    metrics["Traded"],
    metrics["Retired"]
]

plt.bar(
    labels,
    values
)

plt.title(
    "Carbon Credit Supply Metrics"
)

plt.ylabel(
    "Tonnes CO2e"
)

plt.xticks(
    rotation=25
)


# ------------------------------------------------
# Lifecycle Chart
# ------------------------------------------------

plt.subplot(2, 2, 2)

lifecycle = [
    "ISSUED",
    "LISTED",
    "PURCHASED",
    "RETIRED"
]

lifecycle_values = [
    1,
    1,
    1,
    1
]

plt.plot(
    lifecycle,
    lifecycle_values,
    marker="o",
    linewidth=3
)

plt.title(
    "Carbon Credit Lifecycle"
)

plt.ylabel(
    "Transaction Stage"
)

plt.grid(
    True,
    alpha=0.3
)


# ------------------------------------------------
# Security Chart
# ------------------------------------------------

plt.subplot(2, 2, 3)

passed = sum(
    security_df["Result"] == "PASSED"
)

failed = sum(
    security_df["Result"] == "FAILED"
)

plt.bar(
    ["PASSED", "FAILED"],
    [passed, failed]
)

plt.title(
    "Smart Contract Security Tests"
)

plt.ylabel(
    "Number of Tests"
)


# ------------------------------------------------
# Project Status
# ------------------------------------------------

plt.subplot(2, 2, 4)

plt.axis("off")

status_text = f"""

PROJECT STATUS

✓ Issuer Registered

✓ Carbon Credit Issued

✓ Ownership Recorded

✓ Marketplace Listing

✓ Test Purchase

✓ Ownership Transferred

✓ Credit Retired

✓ Retirement Made Irreversible

✓ Security Tests Passed

--------------------------------

Credit ID:
CC-2026-001

CO2e:
10 tCO2e

Final Owner:
{BUYER}

Final Status:
RETIRED

Transactions:
{len(blockchain.transactions)}

Events:
{len(blockchain.events)}

Verification:
SIMULATED ONLY
"""

plt.text(
    0.05,
    0.95,
    status_text,
    verticalalignment="top",
    fontsize=11,
    family="monospace"
)


plt.tight_layout(
    rect=[0, 0, 1, 0.95]
)


# ================================================================
# 17. SAVE OUTPUT AS IMAGE
# ================================================================

output_file = (
    "/content/"
    "carbon_credit_blockchain_result.png"
)

plt.savefig(
    output_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 18. DISPLAY GENERATED IMAGE
# ================================================================

print("\n" + "=" * 75)
print("FINAL OUTPUT IMAGE GENERATED")
print("=" * 75)

print(
    "Saved as:",
    output_file
)

display(
    Image(
        filename=output_file
    )
)


# ================================================================
# 19. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 75)
print("PROJECT EXECUTION COMPLETED SUCCESSFULLY")
print("=" * 75)

print("""
Blockchain Simulation Result:

1. Authorized issuer registered
2. Simulated carbon credit issued
3. Credit assigned to seller wallet
4. Credit listed in marketplace
5. Buyer purchased the credit
6. Ownership transferred
7. Buyer retired the credit
8. Retired credit became non-transferable
9. Retired credit became non-tradable
10. Security tests executed
11. Blockchain event log generated
12. Transaction hashes generated
13. Supply metrics calculated
14. Final dashboard saved as PNG

IMPORTANT:
This is an educational blockchain simulation.
The carbon credit data is simulated and does NOT
represent legally verified or officially issued
carbon credits.
""")

print(
    "\nDownload the generated image from:",
    output_file
)
```
